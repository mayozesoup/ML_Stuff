Neuron Imlpementation

```python
# Numpy, used to do higher level math
import numpy as np

# Sigmoid logistics function
def sigmoid(x):
    out = 1 / (1 + np.exp(-x))
    return out
def d_sigmoid(x):
    return sigmoid(x) * (1 - sigmoid(x))

def mse_loss(y_true, y_pred):
    return ((y_true - y_pred) ** 2).mean()

# Example Neuron
class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def feed_forward(self, inputs):
        total = np.dot(inputs, self.weights) + self.bias
        return sigmoid(total)
```

Simple NeuroNetwork Implementationm

```python
import numpy as np

# Neuro network
class NeuralNetwork:
    """
    1 input layer: 2 inputs
    1 hidden layer: 2 neurons
    1 output layer: 1 neuron
    same weights and bias for all layers
    w: weights
    b: bias
    """
    def __init__(self):
        self.weights = np.array([0,1])
        self.bias = 0

        self.h1 = Neuron(self.weights, self.bias)
        self.h2 = Neuron(self.weights, self.bias)
        self.o1 = Neuron(self.weights, self.bias)

    def feed_forward(self, inputs):
        hidden_out = np.array([self.h1.feed_forward(inputs), self.h2.feed_forward(inputs)])
        out = self.o1.feed_forward(hidden_out)
        return out
```

NeuralNetwork with training

```python
import numpy as np

# Neuro network
class NeuralNetwork:
    """
    1 input layer: 2 inputs
    1 hidden layer: 2 neurons
    1 output layer: 1 neuron
    same weights and bias for all layers
    w: weights
    b: bias
    """
    def __init__(self):
        self.h1 = Neuron(np.random.normal(size=2), np.random.normal())
        self.h2 = Neuron(np.random.normal(size=2), np.random.normal())
        self.o1 = Neuron(np.random.normal(size=2), np.random.normal())

    def feed_forward(self, inputs):
        hidden_out = np.array([self.h1.feed_forward(inputs), self.h2.feed_forward(inputs)])
        out = self.o1.feed_forward(hidden_out)
        return out

    # Training
    def train(self, data, targets, type='sgd'):
        # Inputs are the X inputs
        # Targets are the Y's expected from the X
        if type == "sgd":
            self.__sgd_train(data, targets)
        # TODO implement batch training
        # elif type == 'batch':
        #     self.__batch_train(data, targets)
        else:
            raise ValueError("Invalid training type")

    # Training using SGD
    def __sgd_train(self, data, targets):
        print("SGD Training")

        epoch = 10000
        training_rate = 0.1

        for i in range(epoch):
            for x, y in zip(data, targets):
                h1_sum = np.dot(x, self.h1.weights) + self.h1.bias
                h1_feed = sigmoid(h1_sum)

                h2_sum = np.dot(x, self.h2.weights) + self.h2.bias
                h2_feed = sigmoid(h2_sum)

                o1_sum = np.dot(np.array([h1_feed, h2_feed]), self.o1.weights) + self.o1.bias
                o1_feed = sigmoid(o1_sum)

                # y_predicted
                y_pred = o1_feed

                # partial derivatives
                # loss respect to ypred
                d_l_d_ypred = -2 * (y - o1_feed)

                # o1 neuron w/b's
                d_ypred_d_w5 = d_sigmoid(o1_sum) * h1_feed
                d_ypred_d_w6 = d_sigmoid(o1_sum) * h2_feed
                d_ypred_d_b3 = d_sigmoid(o1_sum)

                # h1/h2 with respect to ypred
                d_ypred_d_h1 = d_sigmoid(o1_sum) * self.o1.weights[0]
                d_ypred_d_h2 = d_sigmoid(o1_sum) * self.o1.weights[1]

                # h1 neuron w/b's
                d_h1_d_w1 = d_sigmoid(h1_sum) * x[0]
                d_h1_d_w2 = d_sigmoid(h1_sum) * x[1]
                d_h1_d_b1 = d_sigmoid(h1_sum)

                # h2 neuron w/b's
                d_h2_d_w3 = d_sigmoid(h2_sum) * x[0]
                d_h2_d_w4 = d_sigmoid(h2_sum) * x[1]
                d_h2_d_b2 = d_sigmoid(h2_sum)

                # Update weights/biases
                self.h1.weights[0] -= training_rate * d_l_d_ypred * d_ypred_d_h1 * d_h1_d_w1
                self.h1.weights[1] -= training_rate * d_l_d_ypred * d_ypred_d_h1 * d_h1_d_w2
                self.h1.bias -= training_rate * d_l_d_ypred * d_ypred_d_h1 * d_h1_d_b1

                self.h2.weights[0] -= training_rate * d_l_d_ypred * d_ypred_d_h2 * d_h2_d_w3
                self.h2.weights[1] -= training_rate * d_l_d_ypred * d_ypred_d_h2 * d_h2_d_w4
                self.h2.bias -= training_rate * d_l_d_ypred * d_ypred_d_h2 * d_h2_d_b2

                self.o1.weights[0] -= training_rate * d_l_d_ypred * d_ypred_d_w5
                self.o1.weights[1] -= training_rate * d_l_d_ypred * d_ypred_d_w6
                self.o1.bias -= training_rate * d_l_d_ypred * d_ypred_d_b3



            if i % 10 == 0:
                y_preds = np.apply_along_axis(self.feed_forward, 1, data)
                loss = mse_loss(targets, y_preds)
                print("Epoch %d loss: %.6f" % (i + 1, loss))
```
